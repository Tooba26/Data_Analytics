```python
from google.cloud import bigquery as bq
```

Step 1: Create Client object

**Client** objects hold projects and a connection to the BigQuery service

```python
client = bq.Client()
```

Step 2: Use DatasetReference("project", "dataset_id)

Project: Collection of datasets

dataset: Collection of tables

```python
dataset_ref = bq.DatasetReference("bigquery-public-data", "hacker_news")
```

Step 3: Use get_dataset() to fetch data

```python
dataset = client.get_dataset(dataset_ref)
```

Use list_tables() to list the tables in dataset

```python
# List all the tables in the dataset
tables = list(client.list_tables(dataset))
```

```python
# Print names of all tables in dataset
for table in tables:
  print(table.table_id)
```

Fetch a table from dataset (if there are multiple tables)

Use dataset_ref() for reference to the table

Use get_table() to fetch the table

```python
# Construct a reference to the desired table
table_ref = dataset_ref.table("full")
```

```python
# API call to fetch the table
table = client.get_table(table_ref)
```

```python
print(table)
```

# Table Schema

Schema: Structure of a table

Each SchemaField shows a specific column(field)

SchemaField('name', 'field type', 'mode (default:NULLABLE)', description of the data, ())

**NOTE:** Mode is by default NULLABLE. It means that a column allows NULL values.

```python
# Print info on all the columns of the fetched table
table.schema

```

**Preview** Some Table Data

Use list_rows(table_name, max_results)

```python
# Preview the first five lines of the table
client.list_rows(table, max_results=50).to_dataframe()
```

```python
# Preview the Fist five entries in the specific column of the table
client.list_rows(table, selected_fields=table.schema[:1], max_results=5).to_dataframe()
```
